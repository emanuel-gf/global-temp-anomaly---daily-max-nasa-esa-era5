```python
import time
from datetime import datetime, timezone
import numpy as np
import requests
import pandas as pd
reference_minutute = np.array([20,50])

import numpy as np 
def get_metar_data(id="EGLC"):
    response = requests.get(
        'https://aviationweather.gov/api/data/metar',
        params={'ids': id, 'format': 'json'}
    )
    data = response.json()
    return data

def make_df_metar(data):
    df = pd.DataFrame().from_dict(data)

    ## convert the time of the observation to dateitme
    df['obsTime'] = df['obsTime'].apply(lambda x: datetime.fromtimestamp(int(x),
                                                                            tz = timezone.utc))

    ## filter only the selected rows
    df = df[['icaoId','obsTime','reportTime','temp','dewp','wdir','wspd','qcField']]

    return df

## GET TIME NOW 
actual = datetime.now()
print(f"TIME NOW:{actual.strftime('%H:%M:%S')}")
actual_minute =actual.minute
correct_min_evaluate = reference_minutute[np.argmin(np.abs(reference_minutute - actual_minute))]

## FIRST API CALL 
data = get_metar_data()
obstime = datetime.fromtimestamp(int(data[0].get('obsTime')),tz = timezone.utc)

while obstime.minute != correct_min_evaluate:
    print(f"Waiting for the right minute: {correct_min_evaluate} - actual minute: {obstime.minute}")
    data = get_metar_data()
    obstime = datetime.fromtimestamp(int(data[0].get('obsTime')),tz = timezone.utc)
    time.sleep(15)

## pass to a df 
df = make_df_metar(data)
df.head()
```
