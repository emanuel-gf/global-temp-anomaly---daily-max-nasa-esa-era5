# Execution
Join the daily parquet files that were excraped from the WeatherUnderground website. 

```python
import pandas as pd 
import fastparquet
import os
from pathlib import Path
import re
```

```python
def clean_df(df, file_stem):
    """
    Clean the dataframes scraped. It eliminates things like: °F °% and convert to a proper dtype

    Args:
        file_stem:Path
             Should contain the date in the name of the file. 
    """
    ## extract the date from the file 
    date_lit = file_stem.strip().split('_',1)[1].replace('_','-')
    
    ## dropnan
    df = df.dropna().copy()

    ## convert to datetime format
    df['datetime'] = pd.to_datetime(date_lit + ' ' + df['Time'], format='%Y-%m-%d %I:%M %p')

    strip_num = lambda x: float(re.sub(r'[^\d.]', '', x)) if pd.notna(x) else None

    df['Temperature'] = df['Temperature'].apply(strip_num)
    df['Dew Point']   = df['Dew Point'].apply(strip_num)
    df['Humidity']    = df['Humidity'].apply(strip_num)
    df['Wind Speed']  = df['Wind Speed'].apply(strip_num)
    df['Wind Gust']   = df['Wind Gust'].apply(strip_num)
    df['Pressure']    = df['Pressure'].apply(strip_num)
    df['Precip.']     = df['Precip.'].apply(strip_num)

    return df
```

```python
## root of the year directory
root = "/home/camarada/Documents/projects/temp-grss-nasa/data_/wunderground"

## where to save the new dir
root_save_processed = "/home/camarada/Documents/projects/temp-grss-nasa/data_/wunderground/processed"
os.makedirs(root_save_processed, exist_ok=True)

year_folder_list = [2021,2022]
for year in year_folder_list:
    print("-"*30)
    print(f"Working year: {year}")
    pathh = os.path.join(root,str(year))
    
    ## create a dir to save the processed
    root_save_processed_year = os.path.join(root_save_processed, str(year))
    os.makedirs(root_save_processed_year, exist_ok=True)

    ## LOOP
    ## loop through all the dir and modify it saving one parquet per month 
    for dirpath, dirs, files in os.walk(pathh):
        for dir in dirs:
            dir_path = os.path.join(pathh, dir)
            print(f"Working dir: {dir}")

            ## create a new dir to save it 
            root_save_processed_year_month = os.path.join(root_save_processed_year, dir)
            os.makedirs(root_save_processed_year_month, exist_ok=True)
            ## list all files inside the dir
            for file in Path(dir_path).iterdir():
                if file.is_file():
                    print(f"Pre-processing file: {file.stem}")
                    df_r = pd.read_parquet(file)
                    ## preprocess
                    df = clean_df(df_r, file.stem)

                    ## save it as parquet
                    save_path = os.path.join(root_save_processed_year_month, file.name)
                    df.to_parquet(save_path,
                                  index=False
                    )
                    print(f"parquet saved at: {save_path}")

                

            
```
